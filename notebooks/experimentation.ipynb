# Aspect-Based Sentiment Analysis (ABSA) of Product Reviews
## Academic Laboratory Experimentation & Benchmark Notebook

This notebook demonstrates the end-to-end NLP and Machine Learning pipeline for:
1. **Text Preprocessing** (Contraction expansion, tokenization, negation preservation, lemmatization)
2. **Aspect Extraction** (Dependency parsing, clause segmentation, taxonomy mapping)
3. **Model Training & Benchmarking** (Logistic Regression, Multinomial Naive Bayes, VADER)
4. **Evaluation & Confusion Matrix Analysis**

```python
import sys
from pathlib import Path

# Add root project directory to system path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
from utils.dataset_loader import load_dataset
from preprocessing.text_preprocessing import TextPreprocessor
from models.aspect_extractor import AspectExtractor
from models.sentiment_model import HybridSentimentAnalyzer
from models.model_evaluation import evaluate_models
```

### Step 1: Load and Inspect Dataset

```python
df = load_dataset()
print(f"Total reviews loaded: {len(df)}")
df.head()
```

### Step 2: Test Text Preprocessing Pipeline

```python
preprocessor = TextPreprocessor()
sample_review = "The phone camera is excellent, battery backup is poor, and the display quality is amazing."

stages = preprocessor.get_preprocessing_stages(sample_review)
for stage, output in stages.items():
    print(f"{stage:25}: {output}")
```

### Step 3: Test Aspect Extraction and Clause Segmentation

```python
aspect_extractor = AspectExtractor()
aspects = aspect_extractor.extract_aspect_contexts(sample_review)

for a in aspects:
    print(f"Aspect: {a['aspect']:12} | Term: {a['raw_term']:15} | Context: {a['context']}")
```

### Step 4: Model Training and Empirical Evaluation

```python
results = evaluate_models(force_retrain=True)

print(f"Logistic Regression Accuracy : {results['logistic_regression']['accuracy'] * 100:.2f}%")
print(f"Naive Bayes Accuracy         : {results['naive_bayes']['accuracy'] * 100:.2f}%")
print(f"VADER Baseline Accuracy      : {results['vader']['accuracy'] * 100:.2f}%")
print(f"Aspect Extraction F1-Score   : {results['aspect_extraction']['f1_score'] * 100:.2f}%")
```

### Step 5: End-to-End Aspect-Based Sentiment Analysis (ABSA)

```python
analyzer = HybridSentimentAnalyzer()
output = analyzer.analyze_review_absa(sample_review, model_name="logistic_regression")

print("=== ABSA OUTPUT ===")
print(f"Review: {output['original_text']}")
print(f"Overall Sentiment: {output['overall_sentiment']}")
print("\nAspect-Level Sentiments:")
for asp in output["aspects"]:
    print(f"  * {asp['aspect']:12}: {asp['sentiment']:8} (Confidence: {asp['confidence']*100:.1f}%) [Context: {asp['context_clause']}]")
```
